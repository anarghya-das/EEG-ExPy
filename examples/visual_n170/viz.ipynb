```python
# Define data paths more robustly
eegnb_data_path = os.path.join(os.path.expanduser('~'), '.eegnb', 'data')
# Fix the folder name case - it should be 'visual_n170' not 'visual_N170'
n170_data_path = os.path.join(eegnb_data_path, 'visual_n170', 'local', 'lsl', 'subject0001', 'session000')

# Check if the directory exists and list available files
if os.path.exists(n170_data_path):
    print(f"Data directory found: {n170_data_path}")
    csv_files = [f for f in os.listdir(n170_data_path) if f.endswith('.csv')]
    print(f"Available CSV files: {csv_files}")
    
    # Use the most recent file or specify one
    if csv_files:
        fn = csv_files[-1]  # Use the last (most recent) file
        print(f"Using file: {fn}")
        
        # Load the data
        data_file_path = os.path.join(n170_data_path, fn)
        df = pd.read_csv(data_file_path)
        print(f"Loaded data shape: {df.shape}")
        print(f"Columns: {list(df.columns)}")
        df.head()
    else:
        print("No CSV files found in the directory")
else:
    print(f"Directory not found: {n170_data_path}")
    print("Available directories:")
    base_path = os.path.join(eegnb_data_path, 'visual_n170', 'local', 'lsl')
    if os.path.exists(base_path):
        for item in os.listdir(base_path):
            print(f"  {item}")
```
